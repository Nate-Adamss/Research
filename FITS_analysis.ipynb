# FITS file analysis

neo7unwise-2668p348-w1-img-m.fits is an unWISE coadd image of NGC 6447 from the W1 band, retrieved from the unWISE archive. The aperture radius is 2.1818 pixels based on WISE's 2.75"/pixel plate scale. 6 arcseconds is the radius of NGC 6447's central region

```python
from astropy.io import fits
from astropy.wcs import WCS
from astropy.coordinates import SkyCoord
import astropy.units as u
from photutils.aperture import CircularAperture, CircularAnnulus, aperture_photometry
import numpy as np
```

```python
# Load the image
hdu = fits.open('neo7unwise-2668p348-w1-img-m.fits')
image = hdu[0].data
wcs = WCS(hdu[0].header)

# Define your target position in RA/Dec
coords = SkyCoord(ra=266.57175*u.degree, dec=35.57192*u.degree)

# Convert sky coords to pixel coords
x_pix, y_pix = wcs.world_to_pixel(coords)
position = (x_pix, y_pix)
```

```python
# Central aperture (e.g. 6" radius ~ 1 PSF FWHM for WISE W1)
# First convert arcsec to pixels: 6" / 2.75"/pixel ≈ 2.1818 pixels
central_aperture = CircularAperture(position, r=2.1818)  # radius in pixels

# Outer annulus (inner=2.1818 pixels, outer=4 pixels)
outer_aperture = CircularAnnulus(position, r_in=2.1818, r_out=4.3636)

# Need to have concentric circles. 4 pixels might be the max radius. 2.1818 is the radius observed by the WISE telescope.
```

```python
# Measure flux in each aperture
phot_central = aperture_photometry(image, central_aperture)
phot_outer = aperture_photometry(image, outer_aperture)

central_flux = phot_central['aperture_sum'][0]
outer_flux = phot_outer['aperture_sum'][0]

print(f"Central flux: {central_flux}")
print(f"Outer annulus flux: {outer_flux}")
print(f"Ratio central/outer: {central_flux/outer_flux:.3f}")
```
